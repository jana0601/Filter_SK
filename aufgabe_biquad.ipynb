## Hinweis
$$
\omega_0 = \frac{2\pi f_0}{f_s}
$$

$$
\alpha = \frac{\sin(\omega_0)}{2Q}
$$

Der Normierungsfaktor lautet:

$$
a_0 = 1 + \alpha
$$

Die normierten Koeffizienten sind:

$$
a_1 = \frac{-2\cos(\omega_0)}{a_0},
\qquad
a_2 = \frac{1-\alpha}{a_0}
$$

$$
b_0 = \frac{1-\cos(\omega_0)}{2a_0},
\qquad
b_1 = 2b_0,
\qquad
b_2 = b_0
$$
